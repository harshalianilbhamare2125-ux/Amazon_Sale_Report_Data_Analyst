```python

import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("Amazon Sale Report.csv")
df.columns = df.columns.str.strip()

fig, ax = plt.subplots(1, 2, figsize=(12, 4))

df['Status'].value_counts().plot(kind='bar', ax=ax[0], color=['#2ecc71','#e74c3c','#f39c12'])
ax[0].set_title("Order Status")

df['Fulfilment'].value_counts().plot(kind='pie', ax=ax[1], autopct='%1.1f%%')
ax[1].set_title("Fulfilment Type")

plt.tight_layout()
plt.show()
```
